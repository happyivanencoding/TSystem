```python
from pathlib import Path

import pandas as pd
from Codes.BacktestEngine import *

from tp_core.data_sources import RETURNS_PATH as CANONICAL_RETURNS_PATH
from tp_core.data_sources import SCREEN_AGGREGATE_PATH

path_screen = str(SCREEN_AGGREGATE_PATH)
path_returns = str(CANONICAL_RETURNS_PATH)
list_noire_path = r"\\groupe-ufg.com\Commun\Prive\GestionAM\Ingenierie_Financiere\PROD\_BASE\_ ESG DATA\Liste_Noire_Exclusion.xlsx"


def run_ptf_backtest(
    name,
    bench,
    percentile,
    esg_exclusion,
    ptf_name,
    ponderation="Market cap",
    top_mandatory=None,
    cap_weight_threshold=None,
    cut_mkt_cap=None,
    liste_noire=None,
    score_neutral="ICB 19",
    weight_neutral="ICB 19",
    backtest_col_sector=None,
    sector_neutral=False,
    start_date_floor=None,
    screen_start_date="mois_impair",
    **ptf_builder_extra,
):
    # 统一从更新后的 screen_aggregate 读取，再生成组合和回测
    # ptf_builder_extra：传给 PtfBuilder 的额外参数，如 country_sector_optimize、top_mandatory_by_country 等
    screen_agg_with_score = pd.read_parquet(path_screen)
    returns = pd.read_parquet(path_returns)
    screen_agg_with_score["Date"] = pd.to_datetime(screen_agg_with_score["Date"])
    returns.index = pd.to_datetime(returns.index)

    bench_col = f"Weight in {bench}"
    if bench_col not in screen_agg_with_score.columns:
        raise ValueError(f"screen_aggregate.parquet 中缺少列: {bench_col}")

    # 过滤掉无法在 returns 中找到足够未来交易日的最后几个月，避免回测尾部越界
    return_dates = pd.Index(returns.index.unique()).sort_values()
    return_dates_set = set(return_dates)

    def has_enough_future_returns(date):
        rebalance_date = date + pd.offsets.MonthBegin(1)
        future_dates = return_dates[return_dates > rebalance_date]
        min_future_count = 1 if rebalance_date in return_dates_set else 2
        return len(future_dates) >= min_future_count

    valid_dates = pd.Index(screen_agg_with_score["Date"].drop_duplicates())
    valid_dates = valid_dates[valid_dates.map(has_enough_future_returns)]
    screen_agg_with_score = screen_agg_with_score[
        screen_agg_with_score["Date"].isin(valid_dates)
    ].copy()

    # 只保留该 benchmark 真正存在的日期，避免稀疏历史导致空组合。
    bench_dates = pd.Index(
        screen_agg_with_score.loc[
            screen_agg_with_score[bench_col].fillna(0) > 0, "Date"
        ].drop_duplicates()
    ).sort_values()
    screen_agg_with_score = screen_agg_with_score[
        screen_agg_with_score["Date"].isin(bench_dates)
    ].copy()

    # 回测起点按该 benchmark 自己首次有 Score ML 的日期确定。
    start_date = screen_agg_with_score.loc[
        (screen_agg_with_score[bench_col].fillna(0) > 0)
        & screen_agg_with_score["Score ML"].notna(),
        "Date"
    ].min()
    if pd.isna(start_date):
        raise ValueError(f"{bench} 中没有可用的 Score ML。")

    # 与 generic_histo_seclist(start_date=...) 一致：可选下限，不早于 ML/业务约定的连续数据起点
    if start_date_floor is not None:
        start_date = max(start_date, pd.to_datetime(start_date_floor))

    ptf_save_path = rf"Portfolio_BT\PTF_ESG_{name}.parquet"
    perf_save_path = rf"Portfolio_BT\Perf_ESG_{name}.parquet"
    list_exclusion_save_path = rf"Portfolio_BT\list_exclusion_{name}.parquet"

    builder = PtfBuilder(
        screen_agg_with_score,
        returns,
        bench=bench,
        ptf_name=ptf_name,
        percentile=percentile,
        esg_exclusion=esg_exclusion,
        liste_noire=liste_noire,
        metrics="Score ML",
        ponderation=ponderation,
        top_mandatory=top_mandatory,
        cap_weight_threshold=cap_weight_threshold,
        cut_mkt_cap=cut_mkt_cap,
        score_neutral=score_neutral,
        weight_neutral=weight_neutral,
        **ptf_builder_extra,
    )

    builder.generic_histo_seclist(
        start_date=start_date,
        freq_rebal=1,
        screen_start_date=screen_start_date,
    )

    # 与 score/weight 中性化一致的分组列；若指定则先跑回测以便 backtest_plot_ptf_bench 复用 perf_ptf
    if backtest_col_sector is not None:
        builder.backtest(
            builder.sec_list_historical,
            col_sector=backtest_col_sector,
            sector_neutral=sector_neutral,
        )

    builder.sec_list_historical.to_parquet(ptf_save_path)
    if builder.list_exclusion_histo is not None:
        builder.list_exclusion_histo.to_parquet(list_exclusion_save_path)

    builder.backtest_plot_ptf_bench(save_path=False)
    if builder.perf_ptf is not None:
        perf_to_save = builder.perf_ptf
        if isinstance(perf_to_save, pd.Series):
            perf_to_save = perf_to_save.rename("Perf PTF").to_frame()
        perf_to_save.to_parquet(perf_save_path)

    print(f"{name} backtest saved to {ptf_save_path} and {perf_save_path}")
    return builder
```
