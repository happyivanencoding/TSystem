```python
import pandas as pd
from BacktestEngine import *
from datetime import datetime

path_screen = r"C:\GoogleDrive\TP\screen\screen_aggregate.pkl"
path_returns = r"C:\GoogleDrive\TP\screen\returns.pkl"

screen_agg = pd.read_pickle(path_screen)
returns = pd.read_pickle(path_returns) 
```

```python
screen_agg.filter(like='Percentile').columns.tolist()
```

```python
################ EUROPE ################
bench = 'STOXX EUROPE 600' #'STOXX EUROPE 600'
for metric in [ 'Quality Avg Percentile']:
    metrics = metric
    percentile = 0.05 # Top 600 * 0.10 = 60 boites
    ptf_name = 'ML EU Q1'
    ponderation = 'Market cap' #'Market cap'
    top_mandatory = None
    cap_weight_threshold = None
    cut_mkt_cap = 0

    builder_EU = PtfBuilder(screen_agg, 
                        returns, 
                        bench=bench,
                        ptf_name=ptf_name, 
                        percentile=percentile,
                        liste_noire=None,
                        metrics=metrics, 
                        ponderation = ponderation)

    builder_EU.generic_histo_seclist(start_date=datetime(2019,12,31))
    builder_EU.backtest_plot_ptf_bench(title=f"Backtest {bench} for {metrics} for {percentile*100}th percentile", save_path=None)
```
